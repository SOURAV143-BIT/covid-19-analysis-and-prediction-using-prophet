```python
##prophet
from prophet import Prophet
```

```python
## we will use prophet for forecasting
confirmed_cases_prophet=df.groupby('Date')['Confirmed'].sum().reset_index()
confirmed_cases_prophet
```

```python
# We need to rename the column as ds(datestamp) and y(value)
prophet_data = confirmed_cases_prophet.rename(columns={"Date" : "ds", "Confirmed" : "y"})
prophet_data
```

```python
model=Prophet()
model.fit(prophet_data)
```

```python
future = model.make_future_dataframe(periods=7)       # For future 7 days
future
```

```python
forecast = model.predict(future)
forecast
```

```python
forecast[["yhat_lower", "yhat_upper", "yhat"]]
```

```python
model.plot(forecast)
plt.show()
```
